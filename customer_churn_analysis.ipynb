Customer Churn Driver Analysis
This section analyzes the relationship between customer characteristics and churn behavior, including monthly charges, tenure, payment methods, demographics, and internet service types.

Key Findings So Far
Finding 1: Overall Churn Rate
Out of 7,043 customers, 1,869 customers have churned while 5,174 customers remained.

The overall churn rate is approximately 26.5%.

Finding 2: Contract Type Impact
Month-to-Month customers show the highest churn rate.

Customers on One-Year and Two-Year contracts are much more likely to stay with the company.

Long-term contracts significantly improve customer retention.

Project Conclusion
This customer churn analysis identified several factors associated with customer attrition.

Key Findings
The overall churn rate is approximately 26.5%.

Customers with Month-to-Month contracts have the highest churn rate.

Long-term contracts significantly reduce churn.

Customers with shorter tenure are more likely to leave.

Service and pricing factors influence customer retention.

Business Recommendations
Encourage customers to move from Month-to-Month to annual contracts.
Improve onboarding for new customers.
Offer targeted retention programs to high-risk customer segments.
Monitor service plans associated with higher churn rates.
This analysis can help businesses improve customer retention and reduce revenue loss caused by customer churn.
